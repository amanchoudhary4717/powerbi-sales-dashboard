# 📊 Sales Dashboard (Power BI) – Dataset Exploration

This notebook is for **understanding and exploring** the sales dataset that will be used
to build an interactive dashboard in **Power BI**.

Steps:
1. Load the sales dataset  
2. Basic EDA (shape, types, preview)  
3. Explore sales by date, region, and category  
4. Export `sales_data.csv` for use in Power BI  


```
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (7,4)
plt.rcParams["figure.dpi"] = 120
```

```
df = pd.read_csv("sales_data.csv", parse_dates=["order_date"])
df.head()
```

```
df.info()
```

```
df.describe()
```

```
# Total sales over time
sales_by_date = df.groupby("order_date")["sales"].sum()

plt.figure()
sales_by_date.plot()
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.title("Sales Over Time")
plt.tight_layout()
plt.show()
```

```
# Sales by region
sales_by_region = df.groupby("region")["sales"].sum().sort_values(ascending=False)
sales_by_region.plot(kind="bar")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.title("Sales by Region")
plt.tight_layout()
plt.show()
```

```
# Sales by category
sales_by_cat = df.groupby("category")["sales"].sum().sort_values(ascending=False)
sales_by_cat.plot(kind="bar")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.title("Sales by Category")
plt.tight_layout()
plt.show()
```
